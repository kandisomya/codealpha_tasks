```python

from deep_translator import MyMemoryTranslator
from IPython.display import display
import ipywidgets as widgets

text_input = widgets.Textarea(
    placeholder="Enter the text you want to translate...",
    description="Text:",
    layout=widgets.Layout(width="90%", height="120px")
)

source_language = widgets.Dropdown(
    options=["auto", "english", "telugu", "hindi", "french", "spanish"],
    value="auto",
    description="From:"
)

target_language = widgets.Dropdown(
    options=["english", "telugu", "hindi", "french", "spanish"],
    value="hindi",
    description="To:"
)

translate_button = widgets.Button(
    description="Translate",
    button_style="success"
)

output = widgets.Output()

def translate_text(b):
    with output:
        output.clear_output()
        text = text_input.value.strip()
        source = source_language.value
        target = target_language.value

        if not text:
            print("Please enter some text.")
            return

        try:
            result = MyMemoryTranslator(
                source=source,
                target=target
            ).translate(text)

            print("Translated Text:")
            print(result)

        except Exception as e:
            print("Translation failed:", e)
            print("Please try again later if the service is unavailable.")

translate_button.on_click(translate_text)

display(
    text_input,
    source_language,
    target_language,
    translate_button,
    output
)
```
