```python

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from IPython.display import display
import ipywidgets as widgets

faq_questions = [
    "What is artificial intelligence?",
    "What is AI?",
    "What is machine learning?",
    "What is Python?",
    "What is data science?",
    "What is deep learning?",
    "What is an internship?",
    "How can I learn programming?",
    "What is natural language processing?",
    "What is a chatbot?",
    "What is computer vision?"
]

faq_answers = [
    "Artificial Intelligence (AI) enables computers to perform tasks that normally require human intelligence.",
    "Artificial Intelligence (AI) enables computers to perform tasks that normally require human intelligence.",
    "Machine learning is a branch of AI that helps computers learn patterns from data.",
    "Python is a popular programming language used in AI, web development, and data science.",
    "Data science uses data, statistics, and programming to discover useful information.",
    "Deep learning is a type of machine learning that uses multi-layered neural networks.",
    "An internship provides practical experience in a professional field.",
    "Start with Python basics, practise small programs, and build simple projects.",
    "Natural Language Processing (NLP) helps computers understand and process human language.",
    "A chatbot is a program that communicates with users and responds to their questions.",
    "Computer vision enables computers to analyse images and videos."
]

# Add common variations for better matching
faq_questions.extend([
    "Explain AI",
    "Tell me about artificial intelligence",
    "Define AI",
    "Explain Python programming"
])

faq_answers.extend([
    faq_answers[1],
    faq_answers[0],
    faq_answers[1],
    faq_answers[3]
])

vectorizer = TfidfVectorizer(ngram_range=(1, 2), stop_words="english")
faq_vectors = vectorizer.fit_transform(faq_questions)

question_input = widgets.Text(
    placeholder="Type your question here...",
    description="You:",
    layout=widgets.Layout(width="85%")
)

ask_button = widgets.Button(
    description="Ask",
    button_style="success"
)

chat_output = widgets.Output()

def answer_question(b):
    with chat_output:
        chat_output.clear_output()
        question = question_input.value.strip()

        if not question:
            print("Please enter a question.")
            return

        query_vector = vectorizer.transform([question])
        scores = cosine_similarity(query_vector, faq_vectors)[0]
        best_match = scores.argmax()

        if scores[best_match] < 0.12:
            print("Chatbot: Sorry, I don't know the answer yet.")
        else:
            print("You:", question)
            print("Chatbot:", faq_answers[best_match])

        question_input.value = ""

ask_button.on_click(answer_question)

display(
    widgets.HTML(
        "<h2>🤖 FAQ Chatbot</h2>"
        "<p>Ask me about AI, Python, and technology!</p>"
    ),
    question_input,
    ask_button,
    chat_output
)
```
